# Results Explorer export

Your measurement counts are loaded into the `counts` variable below. Run the cells to explore them.

```python
counts = {
    "1001": 1,
    "1010": 2,
    "1011": 7,
    "1101": 16,
    "1110": 10,
    "1111": 210,
    "0111": 10
}
counts
```

## A few ways to use it

```python
total = sum(counts.values())
probs = {state: n / total for state, n in counts.items()}
top = max(counts, key=counts.get)
print("shots:", total)
print("most likely:", top, counts[top])
probs
```

```python
# If qiskit is available, the standard histogram:
# from qiskit.visualization import plot_histogram
# plot_histogram(counts)

# Otherwise a plain bar chart:
import matplotlib.pyplot as plt
states = sorted(counts)
plt.bar(states, [counts[s] for s in states])
plt.xlabel("state"); plt.ylabel("counts"); plt.xticks(rotation=90)
plt.show()
```
